# Q4: Find the sum and the average of this list of numbers.

Try doing this using a loop.  Then try doing this without using a loop. 

```python
# Found the sum and avg of this list of numbers
my_list = [1, 5, 10, 55, 88, 44, 42, 50, 20, 38]

#With loop:
list_sum_loop = 0
num_elements = 0
for num in my_list:
    list_sum_loop += num
    num_elements += 1
list_avg_loop = list_sum_loop / num_elements

#Without loop:
list_sum = sum(my_list)
list_avg = list_sum / len(my_list)

print(list_sum_loop, list_avg_loop)
print(list_sum, list_avg)
```

# Q5: 
## Write a function that takes a list and returns a new list that has all the duplicates removed.

Example input and expected output:
- input = `["Michele", "Robin", "Sara", "Michele"]`
- expected output = `['Michele', 'Robin', 'Sara']`


```python
# Wrote a function that takes a list and returns a new list that has all the duplicates removed

names = ["Michele", "Robin", "Sara", "Michele"]

# Technically the instructions never specified that order had to be preserved
def remove_dupes_orderless(names_list):
    return list(set(names))

# With order preserved
def remove_dupes_ordered(names_list):
    duplicateless_names = []
    seen = set()
    for name in names:
        if name not in seen:
            duplicateless_names.append(name)
            seen.add(name)

    return(duplicateless_names)

print(remove_dupes_orderless(names))
print(remove_dupes_ordered(names))
```

# Q8: Write a function that takes a number as input and returns the following:
* If the input is divisible by three, return `'fizz'`
* If the input is divisible by five, return `'buzz'`
* If the input is divisible by three and by five, return `'fizzbuzz'`
* If the input is not divisible by three or five, return `None`.

```python
# Wrote fizzbuzz function

def fizzbuzz(numb):
    if numb % 3 == 0 and numb % 5 == 0:
        return ('fizzbuzz')
    elif numb % 3 == 0:
        return('fizz')
    elif numb % 5 == 0:
        return ('buzz')
    else:
         return None
```

```python
# Testing to see if my function works properly.
print(fizzbuzz(3))
print(fizzbuzz(5))
print(fizzbuzz(15))
```
